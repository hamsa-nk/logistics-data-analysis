```python
df_model = df[df["delivery_time_hours"] > 0].copy()

print("Original rows:", len(df))
print("Rows after removing 0-hour records:", len(df_model))
print("Removed rows:", len(df) - len(df_model))
```

```python
df_model = df[df["delivery_time_hours"] > 0].copy()

print("Original rows:", len(df))
print("Rows after cleaning:", len(df_model))
print("Removed rows:", len(df) - len(df_model))
```

```python
features = [
    "delivery_partner",
    "package_type",
    "vehicle_type",
    "delivery_mode",
    "region",
    "weather_condition",
    "distance_km",
    "package_weight_kg"
]

X = df_model[features]
y = df_model["delivery_time_hours"]

print("X shape:", X.shape)
print("y shape:", y.shape)

print("\nFeatures:")
print(X.columns.tolist())

print("\nTarget:")
print(y.name)
```

```python
categorical_features = X.select_dtypes(include=["object"]).columns.tolist()
numerical_features = X.select_dtypes(exclude=["object"]).columns.tolist()

print("Categorical features:")
print(categorical_features)

print("\nNumerical features:")
print(numerical_features)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

preprocessor = ColumnTransformer(
    transformers=[
        (
            "categorical",
            OneHotEncoder(handle_unknown="ignore"),
            categorical_features
        )
    ],
    remainder="passthrough"
)

print("Preprocessor created successfully.")
```

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression

linear_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("model", LinearRegression())
    ]
)

linear_model.fit(X_train, y_train)

linear_predictions = linear_model.predict(X_test)

print("Linear Regression trained successfully.")
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

linear_mae = mean_absolute_error(y_test, linear_predictions)
linear_rmse = np.sqrt(mean_squared_error(y_test, linear_predictions))
linear_r2 = r2_score(y_test, linear_predictions)

print("Linear Regression Results")
print("-------------------------")
print("MAE :", linear_mae)
print("RMSE:", linear_rmse)
print("R²  :", linear_r2)
```

```python
from sklearn.tree import DecisionTreeRegressor

tree_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("model", DecisionTreeRegressor(random_state=42))
    ]
)

tree_model.fit(X_train, y_train)

tree_predictions = tree_model.predict(X_test)

print("Decision Tree trained successfully.")
```

```python
tree_mae = mean_absolute_error(y_test, tree_predictions)
tree_rmse = np.sqrt(mean_squared_error(y_test, tree_predictions))
tree_r2 = r2_score(y_test, tree_predictions)

print("Decision Tree Results")
print("--------------------")
print("MAE :", tree_mae)
print("RMSE:", tree_rmse)
print("R²  :", tree_r2)
```

```python
from sklearn.ensemble import RandomForestRegressor

forest_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "model",
            RandomForestRegressor(
                n_estimators=100,
                random_state=42,
                n_jobs=-1
            )
        )
    ]
)

forest_model.fit(X_train, y_train)

forest_predictions = forest_model.predict(X_test)

print("Random Forest trained successfully.")
```

```python
forest_mae = mean_absolute_error(y_test, forest_predictions)
forest_rmse = np.sqrt(mean_squared_error(y_test, forest_predictions))
forest_r2 = r2_score(y_test, forest_predictions)

print("Random Forest Results")
print("--------------------")
print("MAE :", forest_mae)
print("RMSE:", forest_rmse)
print("R²  :", forest_r2)
```

```python
results = {
    "Model": [
        "Linear Regression",
        "Decision Tree",
        "Random Forest"
    ],
    "MAE": [
        linear_mae,
        tree_mae,
        forest_mae
    ],
    "RMSE": [
        linear_rmse,
        tree_rmse,
        forest_rmse
    ],
    "R2": [
        linear_r2,
        tree_r2,
        forest_r2
    ]
}

results_df = pd.DataFrame(results)

print(results_df)
```

```python
from sklearn.model_selection import cross_val_score

cv_scores = cross_val_score(
    linear_model,
    X,
    y,
    cv=5,
    scoring="r2"
)

print("5-Fold Cross-Validation R² Scores:")
print(cv_scores)

print("Mean CV R²:", cv_scores.mean())
print("Standard Deviation:", cv_scores.std())
```

```python
forest_cv_scores = cross_val_score(
    forest_model,
    X,
    y,
    cv=5,
    scoring="r2"
)

print("5-Fold Cross-Validation R² Scores:")
print(forest_cv_scores)

print("Mean CV R²:", forest_cv_scores.mean())
print("Standard Deviation:", forest_cv_scores.std())
```

```python
cv_results = pd.DataFrame({
    "Model": [
        "Linear Regression",
        "Random Forest"
    ],
    "Mean CV R²": [
        cv_scores.mean(),
        forest_cv_scores.mean()
    ],
    "Std CV R²": [
        cv_scores.std(),
        forest_cv_scores.std()
    ]
})

print(cv_results)
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
param_grid = {
    "model__n_estimators": [50, 100, 150],
    "model__max_depth": [None, 10, 20],
    "model__min_samples_split": [2, 5]
}

print("Parameter grid created successfully.")
```

```python
grid_search = GridSearchCV(
    estimator=forest_model,
    param_grid=param_grid,
    cv=5,
    scoring="r2",
    n_jobs=-1
)

print("GridSearchCV created successfully.")
```

```python
grid_search.fit(X, y)

print("Hyperparameter tuning completed.")
```

```python
print("Best Parameters:")
print(grid_search.best_params_)

print("\nBest Cross-Validation R²:")
print(grid_search.best_score_)
```

```python
best_forest_model = grid_search.best_estimator_

tuned_forest_predictions = best_forest_model.predict(X_test)

tuned_forest_mae = mean_absolute_error(
    y_test,
    tuned_forest_predictions
)

tuned_forest_rmse = np.sqrt(
    mean_squared_error(
        y_test,
        tuned_forest_predictions
    )
)

tuned_forest_r2 = r2_score(
    y_test,
    tuned_forest_predictions
)

print("Tuned Random Forest Results")
print("---------------------------")
print("MAE :", tuned_forest_mae)
print("RMSE:", tuned_forest_rmse)
print("R²  :", tuned_forest_r2)
```

```python
final_results = pd.DataFrame({
    "Model": [
        "Linear Regression",
        "Decision Tree",
        "Random Forest",
        "Tuned Random Forest"
    ],
    "MAE": [
        linear_mae,
        tree_mae,
        forest_mae,
        tuned_forest_mae
    ],
    "RMSE": [
        linear_rmse,
        tree_rmse,
        forest_rmse,
        tuned_forest_rmse
    ],
    "R2": [
        linear_r2,
        tree_r2,
        forest_r2,
        tuned_forest_r2
    ]
})

print(final_results)
```

```python
def get_original_feature(feature):
    if feature.startswith("categorical__"):
        feature = feature.replace("categorical__", "", 1)
        return feature.split("_", 1)[0]
    elif feature.startswith("remainder__"):
        return feature.replace("remainder__", "", 1)
    return feature

feature_importance_df["original_feature"] = (
    feature_importance_df["feature"].apply(get_original_feature)
)

grouped_importance = (
    feature_importance_df
    .groupby("original_feature")["importance"]
    .sum()
    .sort_values(ascending=False)
)

grouped_importance
```

```python
original_features = [
    "delivery_partner",
    "package_type",
    "vehicle_type",
    "delivery_mode",
    "region",
    "weather_condition",
    "distance_km",
    "package_weight_kg"
]

def get_original_feature(feature):
    for original in original_features:
        if original in feature:
            return original
    return feature

feature_importance_df["original_feature"] = (
    feature_importance_df["feature"].apply(get_original_feature)
)

grouped_importance = (
    feature_importance_df
    .groupby("original_feature")["importance"]
    .sum()
    .sort_values(ascending=False)
)

grouped_importance
```
