```python
import pandas as pd

df = pd.read_csv("cleaned_delivery_logistics.csv")

print(df.head())
print(df.shape)
print(df.info())
```

```python
print(df.isnull().sum())
print("Duplicates:", df.duplicated().sum())
```

```python
print("Unique delivery IDs:", df["delivery_id"].nunique())
print(df["delivery_time_hours"].describe())
print("\nDelivery Partners:")
print(df["delivery_partner"].unique())

print("\nVehicle Types:")
print(df["vehicle_type"].unique())

print("\nDelivery Modes:")
print(df["delivery_mode"].unique())

print("\nWeather Conditions:")
print(df["weather_condition"].unique())

```

```python
print(
    df[
        [
            "distance_km",
            "package_weight_kg",
            "delivery_time_hours",
            "expected_time_hours",
            "delivery_cost",
            "delay_hours"
        ]
    ].corr()
)
```

```python
print("Zero-hour deliveries:", (df["delivery_time_hours"] == 0).sum())
print(df["delivery_time_hours"].value_counts().sort_index())
```

```python
print(
    df.groupby("vehicle_type")["delivery_time_hours"]
      .mean()
      .sort_values()
)
```
